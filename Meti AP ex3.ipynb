Assignment 1: Class Design and Encapsulation

Report

The Student class was designed to represent a student in a university system. The class contains attributes for student ID, name, program, and GPA. To demonstrate encapsulation, the GPA attribute was made private and can only be accessed through getter and setter methods. The setter method validates GPA values to ensure they remain between 0.0 and 4.0. The display_info() method provides a formatted summary of student information. This implementation demonstrates class creation, constructors, object instantiation, and data protection through encapsulation.

Test Cases
1. Create a Student object with valid information.
2. Retrieve GPA using the getter method.
3. Update GPA using the setter method.
4. Display student information.
5. Test GPA validation with invalid values.

```python
class Student:
    def __init__(self, student_id, name, program, gpa):
        self.student_id = student_id
        self.name = name
        self.program = program
        self.set_gpa(gpa)  
    def get_gpa(self):
        return self.__gpa

    def set_gpa(self, gpa):
        if 0.0 <= gpa <= 4.0:
            self.__gpa = gpa
        else:
            print("Error: GPA must be between 0.0 and 4.0")

    def display_info(self):
        print("Student Information")
        print("-------------------")
        print(f"Student ID: {self.student_id}")
        print(f"Name: {self.name}")
        print(f"Program: {self.program}")
        print(f"GPA: {self.__gpa}")


student1 = Student("GSR/6764/18", "Meti Habtamu", "Bioinformatics", 3.89)

student1.display_info()

print("\nCurrent GPA:", student1.get_gpa())

student1.set_gpa(3.9)

print("\nUpdated Information:")
student1.display_info()
```

Assignment 2: Inheritance with Animal Classes

Report

This assignment demonstrates inheritance and polymorphism using an Animal parent class and three child classes: Dog, Cow, and Bird. The parent class contains common attributes such as name and age and defines a speak() method. Each child class overrides the speak() method to provide species-specific behavior. Multiple animal objects are stored in a list and processed using a loop. This demonstrates polymorphism because the same method call produces different outputs depending on the object type. The design promotes code reuse and extensibility.

Test Cases
1. Create Dog, Cow, and Bird objects.
2. Store all objects in a list.
3. Call speak() for each object.
4. Verify that each animal produces the correct sound.

```python
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def speak(self):
        print("Some generic animal sound")


class Dog(Animal):
    def speak(self):
        print(f"{self.name} says: Woof! Woof!")


class Cow(Animal):
    def speak(self):
        print(f"{self.name} says: Moo! Moo!")


class Bird(Animal):
    def speak(self):
        print(f"{self.name} says: Tweet! Tweet!")


dog = Dog("Bobby", 3)
cow = Cow("Burre", 5)
bird = Bird("Cok", 2)

animals = [dog, cow, bird]

print("Animal Sounds")
print("-------------")

for animal in animals:
    print(f"Name: {animal.name}, Age: {animal.age}")
    animal.speak()
    print()
```

Assignment 3: Bioinformatics Sequence System

Report

This assignment models biological sequences using object-oriented programming. A parent Sequence class stores common attributes such as sequence ID and sequence data. DNASequence, RNASequence, and ProteinSequence inherit from the parent class and implement specialized functionality. DNA sequences can calculate GC content, RNA sequences can be converted to DNA, and protein sequences can count amino acids. Validation methods ensure that only valid biological characters are accepted. This design demonstrates inheritance, abstraction, and real-world scientific modeling.

Test Cases
1. Create valid DNA, RNA, and Protein sequence objects.
2. Verify sequence length calculation.
3. Calculate GC content for DNA sequences.
4. Convert RNA sequences to DNA.
5. Count amino acids in protein sequences.
6. Test invalid sequence input.

```python
# Parent Class
class Sequence:
    def __init__(self, seq_id, sequence):
        self.seq_id = seq_id
        self.sequence = sequence.upper()

    def get_length(self):
        return len(self.sequence)

    def display_info(self):
        print(f"Sequence ID: {self.seq_id}")
        print(f"Sequence: {self.sequence}")
        print(f"Length: {self.get_length()}")


# DNA Sequence Class
class DNASequence(Sequence):
    def __init__(self, seq_id, sequence):
        super().__init__(seq_id, sequence)

        valid_chars = {'A', 'T', 'G', 'C'}
        if not set(self.sequence).issubset(valid_chars):
            raise ValueError("Invalid DNA sequence! Only A, T, G, and C are allowed.")

    def gc_content(self):
        gc_count = self.sequence.count('G') + self.sequence.count('C')
        return (gc_count / len(self.sequence)) * 100


# RNA Sequence Class
class RNASequence(Sequence):
    def __init__(self, seq_id, sequence):
        super().__init__(seq_id, sequence)

        valid_chars = {'A', 'U', 'G', 'C'}
        if not set(self.sequence).issubset(valid_chars):
            raise ValueError("Invalid RNA sequence! Only A, U, G, and C are allowed.")

    def to_dna(self):
        return self.sequence.replace('U', 'T')


# Protein Sequence Class
class ProteinSequence(Sequence):
    def __init__(self, seq_id, sequence):
        super().__init__(seq_id, sequence)

        valid_chars = set("ACDEFGHIKLMNPQRSTVWY")
        if not set(self.sequence).issubset(valid_chars):
            raise ValueError("Invalid Protein sequence!")

    def count_amino_acids(self):
        amino_acid_counts = {}

        for aa in self.sequence:
            amino_acid_counts[aa] = amino_acid_counts.get(aa, 0) + 1

        return amino_acid_counts


# Testing the Classes
# -------------------------

# DNA Example
dna = DNASequence("DNA001", "ATGCGCGTA")
print("DNA Sequence")
dna.display_info()
print(f"GC Content: {dna.gc_content():.2f}%")
print()

# RNA Example
rna = RNASequence("RNA001", "AUGCGCUAA")
print("RNA Sequence")
rna.display_info()
print("Converted DNA:", rna.to_dna())
print()

# Protein Example
protein = ProteinSequence("PROT001", "MKTLLVAA")
print("Protein Sequence")
protein.display_info()
print("Amino Acid Counts:")
print(protein.count_amino_acids())
```

Assignment 4: Library Management System

Report

The Library Management System demonstrates object relationships and composition. The system consists of Book, Member, and Library classes. The Library class manages collections of books and members and coordinates borrowing and returning operations. Book availability is tracked to prevent multiple users from borrowing the same book simultaneously. Each member maintains a list of borrowed books. This design demonstrates state management, composition, and interaction between multiple objects.

Test Cases
1. Add books to the library.
2. Register new members.
3. Borrow available books.
4. Attempt to borrow unavailable books.
5. Return borrowed books.
6. Verify member borrowing records.

```python
# Book Class
class Book:
    def __init__(self, book_id, title, author):
        self.book_id = book_id
        self.title = title
        self.author = author
        self.available = True

    def __str__(self):
        status = "Available" if self.available else "Borrowed"
        return f"{self.book_id} - {self.title} by {self.author} ({status})"


# Member Class
class Member:
    def __init__(self, member_id, name):
        self.member_id = member_id
        self.name = name
        self.borrowed_books = []

    def __str__(self):
        return f"Member ID: {self.member_id}, Name: {self.name}"


# Library Class
class Library:
    def __init__(self):
        self.books = []
        self.members = []

    # Add a book
    def add_book(self, book):
        self.books.append(book)
        print(f"Book '{book.title}' added successfully.")

    # Register a member
    def register_member(self, member):
        self.members.append(member)
        print(f"Member '{member.name}' registered successfully.")

    # Borrow a book
    def borrow_book(self, member_id, book_id):
        member = next((m for m in self.members if m.member_id == member_id), None)
        book = next((b for b in self.books if b.book_id == book_id), None)

        if member and book:
            if book.available:
                book.available = False
                member.borrowed_books.append(book)
                print(f"{member.name} borrowed '{book.title}'.")
            else:
                print(f"Sorry, '{book.title}' is currently unavailable.")
        else:
            print("Member or Book not found.")

    # Return a book
    def return_book(self, member_id, book_id):
        member = next((m for m in self.members if m.member_id == member_id), None)

        if member:
            for book in member.borrowed_books:
                if book.book_id == book_id:
                    book.available = True
                    member.borrowed_books.remove(book)
                    print(f"{member.name} returned '{book.title}'.")
                    return

            print("This member did not borrow the specified book.")
        else:
            print("Member not found.")

    # Display borrowed books of a member
    def display_member_books(self, member_id):
        member = next((m for m in self.members if m.member_id == member_id), None)

        if member:
            print(f"\nBooks borrowed by {member.name}:")
            if member.borrowed_books:
                for book in member.borrowed_books:
                    print(f"- {book.title}")
            else:
                print("No books borrowed.")
        else:
            print("Member not found.")


# Testing the Library System
# ---------------------------

library = Library()

# Add books
book1 = Book(101, "Python Programming", "John Smith")
book2 = Book(102, "Data Structures", "Jane Doe")

library.add_book(book1)
library.add_book(book2)

# Register members
member1 = Member(1, "Meti")
member2 = Member(2, "Abel")

library.register_member(member1)
library.register_member(member2)

# Borrow books
library.borrow_book(1, 101)
library.borrow_book(2, 101)  # Unavailable book

# Display borrowed books
library.display_member_books(1)

# Return a book
library.return_book(1, 101)

# Borrow again
library.borrow_book(2, 101)

# Display borrowed books
library.display_member_books(2)
```

 Assignment 5: Abstract Base Class and Polymorphism

Report

This assignment uses abstraction through the Shape abstract base class. The abstract methods area() and perimeter() define a common interface that all shapes must implement. Circle, Rectangle, and Triangle inherit from Shape and provide their own implementations of these methods. Shape objects are stored in a list and processed using a single loop. This demonstrates polymorphism because different shape types respond to the same method calls. The design ensures consistency while allowing specialized behavior.

Test Cases
1. Create Circle, Rectangle, and Triangle objects.
2. Calculate area and perimeter for each shape.
3. Store all shapes in a list.
4. Use one loop to display results.
5. Verify calculations against known values

```python
from abc import ABC, abstractmethod
import math

# Abstract Base Class
class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


# Circle Class
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius


# Rectangle Class
class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)


# Triangle Class
class Triangle(Shape):
    def __init__(self, side1, side2, side3):
        self.side1 = side1
        self.side2 = side2
        self.side3 = side3

    def perimeter(self):
        return self.side1 + self.side2 + self.side3

    def area(self):
        # Heron's Formula
        s = self.perimeter() / 2
        return math.sqrt(
            s * (s - self.side1) * (s - self.side2) * (s - self.side3)
        )


# Create Shape Objects
circle = Circle(5)
rectangle = Rectangle(4, 6)
triangle = Triangle(3, 4, 5)

shapes = [circle, rectangle, triangle]

# Polymorphism
for shape in shapes:
    print(f"{shape.__class__.__name__}")
    print(f"Area: {shape.area():.2f}")
    print(f"Perimeter: {shape.perimeter():.2f}")
    print("-" * 20)
```

Assignment 6: Variant Annotation System

Report

This assignment models genomic variants using inheritance. The Variant parent class stores chromosome, position, reference allele, and alternate allele information. SNP, Insertion, and Deletion classes inherit from Variant and override methods to classify variant types and calculate variant lengths. Each variant can generate an annotation summary describing its characteristics. The design reflects real-world bioinformatics applications and demonstrates domain-specific object-oriented modeling.

Test Cases
1. Create SNP, Insertion, and Deletion objects.
2. Verify variant type classification.
3. Calculate variant lengths.
4. Print annotation summaries.
5. Compare outputs across different variant types.

```python
# Parent Class
class Variant:
    def __init__(self, chromosome, position, ref_allele, alt_allele):
        self.chromosome = chromosome
        self.position = position
        self.ref_allele = ref_allele
        self.alt_allele = alt_allele

    def classify_variant(self):
        return "Generic Variant"

    def variant_length(self):
        return abs(len(self.alt_allele) - len(self.ref_allele))

    def annotation_summary(self):
        print("Variant Annotation Summary")
        print("--------------------------")
        print(f"Chromosome: {self.chromosome}")
        print(f"Position: {self.position}")
        print(f"Reference Allele: {self.ref_allele}")
        print(f"Alternate Allele: {self.alt_allele}")
        print(f"Variant Type: {self.classify_variant()}")
        print(f"Variant Length: {self.variant_length()}")
        print()


# SNP Class
class SNP(Variant):
    def classify_variant(self):
        return "Single Nucleotide Polymorphism (SNP)"

    def variant_length(self):
        return 1


# Insertion Class
class Insertion(Variant):
    def classify_variant(self):
        return "Insertion"

    def variant_length(self):
        return len(self.alt_allele) - len(self.ref_allele)


# Deletion Class
class Deletion(Variant):
    def classify_variant(self):
        return "Deletion"

    def variant_length(self):
        return len(self.ref_allele) - len(self.alt_allele)


# Testing the Classes
# -------------------------

snp = SNP("chr1", 1050, "A", "G")
insertion = Insertion("chr2", 2500, "T", "TAC")
deletion = Deletion("chr3", 5000, "GCTA", "G")

variants = [snp, insertion, deletion]

for variant in variants:
    variant.annotation_summary()
```

Assignment 7: Alternative Constructors

Report

The Genome class demonstrates the use of alternative constructors through class methods. The standard constructor accepts a genome name, species, and size. The from_csv_line() class method creates objects from CSV-formatted strings, while from_dict() creates objects from dictionaries. These methods simplify object creation from structured data sources. This design demonstrates constructor flexibility, data parsing, and the use of class methods.

Test Cases
1. Create a Genome object using the normal constructor.
2. Create a Genome object from a CSV string.
3. Create a Genome object from a dictionary.
4. Display genome information.
5. Verify correct parsing of input data.

```python
# Genome Class

class Genome:
    def __init__(self, name, species, size):
        self.name = name
        self.species = species
        self.size = size

    # Alternative constructor from CSV line
    @classmethod
    def from_csv_line(cls, csv_line):
        name, species, size = csv_line.split(",")
        return cls(name.strip(), species.strip(), int(size.strip()))

    # Alternative constructor from dictionary
    @classmethod
    def from_dict(cls, data):
        return cls(
            data["name"],
            data["species"],
            data["size"]
        )

    # Display genome information
    def display_info(self):
        print("Genome Information")
        print("------------------")
        print(f"Name: {self.name}")
        print(f"Species: {self.species}")
        print(f"Genome Size: {self.size} Mb")
        print()


# Testing the Class
# -------------------------

# Using the normal constructor
genome1 = Genome("Human", "Homo sapiens", 3200)

# Using from_csv_line()
csv_data = "NDama,Bos taurus,2800"
genome2 = Genome.from_csv_line(csv_data)

# Using from_dict()
genome_data = {
    "name": "Arabidopsis",
    "species": "Arabidopsis thaliana",
    "size": 135
}
genome3 = Genome.from_dict(genome_data)

# Display all genomes
genome1.display_info()
genome2.display_info()
genome3.display_info()
```

Assignment 8: Static Methods vs Class Methods

Report

This assignment demonstrates the difference between static methods and class methods. The DNAUtils class contains a class variable called valid_bases that stores allowed DNA characters. Static methods are used for utility functions such as sequence validation and GC content calculation because they do not depend on object or class state. The class method update_valid_bases() modifies the shared class variable. This implementation highlights how static methods, class methods, and class variables serve different purposes within a class.

Test Cases
1. Validate a correct DNA sequence.
2. Validate an incorrect DNA sequence.
3. Calculate GC content.
4. Display the current valid bases.
5. Update valid bases using a class method.
6. Validate sequences using the updated rules.

```python
# DNAUtils Class

class DNAUtils:

    # Class Variable
    valid_bases = "ATGC"

    # Static Method: Check if DNA sequence is valid
    @staticmethod
    def is_valid_dna(sequence):
        sequence = sequence.upper()
        return all(base in DNAUtils.valid_bases for base in sequence)

    # Static Method: Calculate GC Content
    @staticmethod
    def gc_content(sequence):
        sequence = sequence.upper()

        if len(sequence) == 0:
            return 0

        gc_count = sequence.count("G") + sequence.count("C")
        return (gc_count / len(sequence)) * 100

    # Class Method: Update valid bases
    @classmethod
    def update_valid_bases(cls, new_bases):
        cls.valid_bases = new_bases


# Testing the Class
# -------------------------

dna_sequence = "ATGCGCTA"

print("Valid DNA:", DNAUtils.is_valid_dna(dna_sequence))
print("GC Content:", round(DNAUtils.gc_content(dna_sequence), 2), "%")

print("\nCurrent Valid Bases:", DNAUtils.valid_bases)

# Update valid bases
DNAUtils.update_valid_bases("ATGCN")

print("Updated Valid Bases:", DNAUtils.valid_bases)

# Test sequence containing N
print("Valid DNA after update:",
      DNAUtils.is_valid_dna("ATGCN"))
```

Assignment 9: Operator Overloading

Report

This assignment demonstrates operator overloading using Python magic methods. The Sequence class implements len(), str(), eq(), and add(). These methods allow Sequence objects to behave like built-in Python objects. Users can calculate sequence length using len(), print readable object information, compare sequences using ==, and concatenate sequences using the + operator. This approach improves code readability and creates a more Pythonic object-oriented design.

Test Cases
1. Create multiple Sequence objects.
2. Use len() to obtain sequence length.
3. Print sequence objects using print().
4. Compare sequences using ==.
5. Concatenate sequences using +.
6. Verify the resulting sequence information.

```python
# Sequence Class

class Sequence:
    def __init__(self, seq_id, sequence):
        self.seq_id = seq_id
        self.sequence = sequence.upper()

    # Returns sequence length
    def __len__(self):
        return len(self.sequence)

    # Returns readable string representation
    def __str__(self):
        return f"Sequence ID: {self.seq_id}, Sequence: {self.sequence}"

    # Compares two sequences
    def __eq__(self, other):
        return self.sequence == other.sequence

    # Joins two sequences
    def __add__(self, other):
        new_sequence = self.sequence + other.sequence
        return Sequence(
            f"{self.seq_id}_{other.seq_id}",
            new_sequence
        )


# Testing the Operators
# -------------------------

seq1 = Sequence("SEQ001", "ATGC")
seq2 = Sequence("SEQ002", "ATGC")
seq3 = Sequence("SEQ003", "GCTA")

# __str__()
print("Sequence Information:")
print(seq1)

# __len__()
print("\nSequence Length:")
print(len(seq1))

# __eq__()
print("\nSequence Comparison:")
print(seq1 == seq2)
print(seq1 == seq3)

# __add__()
print("\nJoined Sequence:")
combined = seq1 + seq3
print(combined)

print("\nLength of Joined Sequence:")
print(len(combined))
```
