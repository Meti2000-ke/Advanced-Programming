# Question 1

Define the following algorithmic design paradigms and give one real-world example for each.

### a) Brute Force

**Definition:**
Brute Force is an algorithmic approach that systematically tries all possible solutions until the correct or best solution is found. It is simple to implement but can be inefficient for large problems.

**Real-world example:**
Trying every possible password combination until the correct password is found.

### b) Divide and Conquer

**Definition:**
Divide and Conquer solves a problem by dividing it into smaller subproblems, solving each subproblem independently, and then combining their solutions to obtain the final result.

**Real-world example:**
Merge Sort divides a large list into smaller lists, sorts them, and then merges the sorted lists.

### c) Greedy Algorithm

**Definition:**
A Greedy Algorithm builds a solution step by step by choosing the best available option at each step, without reconsidering previous choices. It is effective when local optimal choices lead to a global optimal solution.

**Real-world example:**
Making change using the largest available currency denomination first, such as giving 25, 10, 5, and 1-unit coins when making change.

### d) Dynamic Programming

**Definition:**
Dynamic Programming solves complex problems by breaking them into overlapping subproblems, solving each subproblem once, and storing its result for future use.

**Real-world example:**
Calculating the shortest route between locations when the same smaller routes are repeatedly considered.

### e) Backtracking

**Definition:**
Backtracking builds a solution incrementally and abandons a partial solution as soon as it is determined that it cannot lead to a valid solution. It then returns to a previous decision and tries another option.

**Real-world example:**
Solving a Sudoku puzzle by trying possible numbers and going back whenever a choice creates a conflict.

| **Reconsideration**  | Previous choices are generally not changed.            | Previous subproblem results are reused when needed.                                    |

| **Efficiency**       | Usually simpler and faster.                            | May require more time and memory.                                                      |
| **Example**          | Fractional Knapsack, Dijkstra's algorithm.             | 0/1 Knapsack, Fibonacci sequence.                                                      |

### 2. Backtracking vs Branch & Bound

| **Aspect**            | **Backtracking**                                                                    | **Branch & Bound**                                                                         |
| --------------------- | ----------------------------------------------------------------------------------- | ------------------------------------------------------------------------------------------ |
| **Purpose**           | Mainly used to find feasible or valid solutions.                                    | Mainly used to find an optimal solution.                                                   |
| **Search strategy**   | Explores possible solutions and backtracks when a partial solution becomes invalid. | Explores possible solutions and eliminates branches that cannot produce a better solution. |
| **Pruning condition** | Prunes a branch when it violates constraints.                                       | Prunes a branch when its bound shows it cannot improve the current best solution.          |
| **Typical problems**  | Constraint satisfaction problems.                                                   | Optimization problems.                                                                     |
| **Search order**      | Commonly uses depth-first search (DFS).                                             | Can use BFS, DFS, or a priority queue.                                                     |

````python
##Part II: Programming

#Question 4 

import random

# Count recursive calls
recursive_calls = 0


def merge_sort(arr):
    global recursive_calls
    recursive_calls += 1

    # Base case
    if len(arr) <= 1:
        return arr

    # Divide
    mid = len(arr) // 2
    left = arr[:mid]
    right = arr[mid:]

    # Conquer
    left = merge_sort(left)
    right = merge_sort(right)

    # Combine
    return merge(left, right)


def merge(left, right):
    result = []
    i = 0
    j = 0

    # Compare elements and merge in sorted order
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    # Add remaining elements
    result.extend(left[i:])
    result.extend(right[j:])

    return result


# Generate a random list of integers
arr = [random.randint(1, 100) for _ in range(10)]

# Display original array
print("Original array:")
print(arr)

# Apply Merge Sort
sorted_arr = merge_sort(arr)

# Display sorted array
print("\nSorted array:")
print(sorted_arr)

# Display number of recursive calls
print("\nNumber of recursive calls:", recursive_calls)
````

````python
## Part III: Bioinformatics Applications 

# Question 5


def brute_force_motif_search(dna, motif):
    positions = []
    comparisons = 0

    # Check every possible starting position
    for i in range(len(dna) - len(motif) + 1):
        match = True

        # Compare the motif with the DNA sequence
        for j in range(len(motif)):
            comparisons += 1
            if dna[i + j] != motif[j]:
                match = False
                break

        if match:
            positions.append(i)

    return positions, comparisons


# Input
dna = "ATCGTAGCTAGCTAGCTA"
motif = "TAG"

# Search for the motif
positions, comparisons = brute_force_motif_search(dna, motif)

# Output
print("DNA:", dna)
print("Motif:", motif)
print("Positions where the motif occurs:", positions)
print("Number of comparisons made:", comparisons)

# ---------------------------------------------------------
# Explanation: Why is BLAST faster than brute force?
# ---------------------------------------------------------

# The brute-force approach compares the motif with the DNA
# sequence at every possible position, which can require many
# character-by-character comparisons.
#
# BLAST (Basic Local Alignment Search Tool) is faster because
# it uses a seed-and-extend strategy. It first searches for
# short matching words (seeds) and only performs detailed
# alignment around promising matches.
#
# BLAST also uses indexing and scoring methods to avoid
# comparing every possible position.
#
# Therefore, brute force is simple but can be slow for very
# large DNA sequences, while BLAST reduces the search space
# and is much more efficient for large biological databases.
````

# Question 6: Common DNA Motif Discovery Using Backtracking

### Given DNA Sequences

```text
S1 = ATCGTACGATGCA
S2 = GTCGTACCATGGA
S3 = TTCGTACGGTACA
S4 = ATCGTACTATGCG
```

The common motif is:

```text
CGTAC
```

It occurs in all four sequences starting at position 2 using **0-based indexing**.

---

## 1. How a Backtracking Algorithm Searches for the Motif

A backtracking algorithm builds a possible motif one nucleotide at a time.

1. It starts with an empty motif.
2. It tries possible nucleotides: `A`, `C`, `G`, and `T`.
3. After adding a nucleotide, it checks whether the partial motif can occur in all four DNA sequences.
4. If the partial motif is valid, it continues by adding another nucleotide.
5. If the partial motif cannot occur in all sequences, that branch is **pruned**.
6. The algorithm backtracks to the previous position and tries another nucleotide.
7. When a complete motif is found in all sequences, it is reported as a common motif.

For these sequences, the successful path is:

```text
C → CG → CGT → CGTA → CGTAC
```

Therefore, the common motif is **CGTAC**.

---

## 2. Search Tree

A simplified search tree showing the successful path and pruning is:

```text
                         Start

              +------------+------------+
              |            |            |
              A            C            G/T
              |            |             |
            Prune        CG           Prune
                           |
                         CGT
                           |
                         CGTA
                           |
                        CGTAC
                           |
                     ✓ Common motif
```

A more detailed representation of the successful search is:

The tree is simplified because the algorithm does not need to display every possible branch once a branch is shown to be impossible.

---

## 3. Where Does Pruning Occur?

**Pruning occurs whenever a partial motif cannot be found in all four DNA sequences.**

For example:

```text
CA
```

is not present in all four sequences, so the algorithm immediately stops exploring that branch.

Similarly:

```text
CC
CGG
CGTA A
```

cannot lead to the required common motif, so these branches are discarded.

The successful branch is:

```text
C → CG → CGT → CGTA → CGTAC
```

Once `CGTAC` is found in all four sequences, the search can stop if the task is to find one common motif.

---

## 4. Comparison with Brute Force

| **Aspect**        | **Backtracking**                                           | **Brute Force**                                           |
| ----------------- | ---------------------------------------------------------- | --------------------------------------------------------- |
| **Search method** | Builds a candidate motif step-by-step.                     | Tests all possible candidate motifs/positions.            |
| **Pruning**       | Eliminates invalid branches early.                         | Generally does not eliminate branches early.              |
| **Efficiency**    | More efficient when many candidates can be rejected early. | Can require many unnecessary comparisons.                 |
| **Search space**  | Reduces the search space by pruning.                       | Explores a larger search space.                           |
| **Complexity**    | Usually better in practice when pruning is effective.      | Can become expensive for large sequences.                 |
| **Example here**  | Rejects candidates such as `CA` as soon as they fail.      | Continues checking each possible candidate independently. |

### Conclusion

Backtracking is more efficient than a simple brute-force approach because it **abandons invalid candidate motifs as soon as they fail**. For the given sequences, the algorithm follows the successful path:

```text
C → CG → CGT → CGTA → CGTAC
```

and identifies **CGTAC** as the common DNA motif.
