# Question 1: Definitions

### a. String
A **string** is a sequence of characters from a given alphabet. In bioinformatics, a DNA string is a sequence of nucleotides.

**Example:** `ATCGTAC`

---

### b. Prefix
A **prefix** is a substring that occurs at the **beginning** of a string.

**Example:** In `ATCGTAC`, `ATC` is a prefix.

---

### c. Suffix
A **suffix** is a substring that occurs at the **end** of a string.

**Example:** In `ATCGTAC`, `TAC` is a suffix.

---

### d. Substring
A **substring** is a sequence of consecutive characters occurring **within** a string.

**Example:** In `ATCGTAC`, `CGT` is a substring.

---

### e. Pattern
A **pattern** is a specific sequence that we want to find or match within a larger string.

**Example:** Pattern = `CGT` in the string `ATCGTAC`.

---

### f. Alphabet
An **alphabet** is a finite set of characters used to construct strings.

**Example:** The DNA alphabet is:

`Σ = {A, T, C, G}`

---

### g. String Matching
**String matching** is the process of finding whether, and where, a pattern occurs within a larger string.

**Example:** Searching for `CGT` in `ATCGTAC` → **Match found at position 3.**

---

### Summary Example

For the DNA string `ATCGTAC`:

- **String:** `ATCGTAC`
- **Prefix:** `ATC`
- **Suffix:** `TAC`
- **Substring:** `CGT`
- **Pattern:** `CGT`
- **Alphabet:** `{A, T, C, G}`
- **String matching:** Searching for `CGT` in `ATCGTAC` → Match found.
