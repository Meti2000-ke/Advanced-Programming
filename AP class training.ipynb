Part 1: Basic Input and Output

```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))

print(f"Hello {name}! You are {age} years old.")
```

Part 2: Even or Odd Checker

```python
number = int(input("Enter a number: "))

if number % 2 == 0:
    print(f"{number} is even")
else:
    print(f"{number} is odd")
```

Part 3: Loop Practice

```python
for i in range(1, 21):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

Part 4: List Operations 

```python
numbers = [12, 45, 7, 23, 19]

largest = numbers[0]
for num in numbers:
    if num > largest:
        largest = num

smallest = numbers[0]
for num in numbers:
    if num < smallest:
        smallest = num

total = 0
for num in numbers:
    total += num

print(f"Numbers: {numbers}")
print(f"Largest number: {largest}")
print(f"Smallest number: {smallest}")
print(f"Sum of all numbers: {total}")
```

Part 5: Dictionary Task

```python
student = {
    "name": "Abel",
    "age": 20,
    "grade": "A"
}

student["city"] = "Addis Ababa"

student["grade"] = "A+"  # or any new grade like "B", "A-", etc.

print("Student Information:")
for key, value in student.items():
    print(f"{key}: {value}")
```

Part 6: Function 

```python
def calculate_area(length, width):
    """Calculate and return the area of a rectangle"""
    area = length * width
    return area

result = calculate_area(10, 5)
print(f"The area of the rectangle is: {result}")
```

Part 7: Mini Project 

```python
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

operation = input("Enter an operation (+, -, *, /): ")

if operation == "+":
    result = num1 + num2
    print(f"{num1} + {num2} = {result}")
elif operation == "-":
    result = num1 - num2
    print(f"{num1} - {num2} = {result}")
elif operation == "*":
    result = num1 * num2
    print(f"{num1} × {num2} = {result}")
elif operation == "/":
    if num2 != 0:
        result = num1 / num2
        print(f"{num1} ÷ {num2} = {result}")
    else:
        print("Error: Cannot divide by zero!")
else:
    print("Invalid operation! Please use +, -, *, or /")
```
