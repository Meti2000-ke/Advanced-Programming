# Question 1: Differences Among Genome Graph Concepts

## 1. Linear Reference Genome

A **linear reference genome** is a single, fixed DNA sequence used as a standard reference for aligning and comparing sequencing reads.

* Represents the genome as **one continuous sequence**.
* Usually contains **one chosen version** of each genomic region.
* Easy to store, visualize, and use for read alignment.
* Limitation: it may not represent genetic variation that differs from the reference.

**Main purpose:** To provide a standard sequence for genome alignment and comparison.

---

## 2. Assembly Graph

An **assembly graph** represents the possible relationships between DNA sequence fragments obtained during genome sequencing.

* Built from sequencing reads or **k-mers**.
* Nodes represent sequence fragments or k-mers, while edges represent their connections.
* Shows **alternative paths** caused by repeats, sequencing ambiguity, or genetic variation.
* Used to reconstruct a genome from sequencing data.

**Main purpose:** **Genome assembly** — determining how sequencing fragments fit together.

---

## 3. Variation Graph

A **variation graph** represents a reference genome together with **alternative sequences or variants** at specific genomic locations.

* Contains multiple possible paths through the same genomic region.
* Can represent SNPs, insertions, deletions, and other variants.
* Sequencing reads can be aligned to the appropriate path instead of only to one reference sequence.
* Helps reduce **reference bias** during sequence analysis.

**Main purpose:** **Representing genetic variation around a reference genome.**

---

## 4. Genome Graph

A **genome graph** is a general graph-based representation of a genome in which genomic sequences are represented as **nodes and connections between them as edges**.

* Represents genomic regions and their relationships as a graph.
* Can represent alternative genomic structures and sequences.
* The term is broader than **variation graph** and can include different types of genome graphs.
* Useful for representing complex genomic structures that are difficult to represent using a linear sequence.

**Main purpose:** **Representing genome structure and relationships using a graph.**

---

## 5. Pangenome Graph

A **pangenome graph** represents the genomic sequences and genetic diversity of **multiple individuals, strains, or organisms of the same species** in a single graph.

* Combines genomic information from many genomes.
* Contains both **shared sequences** and **population-specific sequences**.
* Represents genetic diversity such as SNPs, insertions, deletions, and structural variations.
* Provides a more complete representation of a species than a single linear reference genome.

**Main purpose:** **Representing the genetic diversity of an entire species or population.**

---

## Quick Comparison

| Concept                     | What it represents                              | Main purpose                                        |
| --------------------------- | ----------------------------------------------- | --------------------------------------------------- |
| **Linear reference genome** | One continuous reference sequence               | Standard sequence for alignment and comparison      |
| **Assembly graph**          | Relationships among sequencing fragments/k-mers | Genome assembly                                     |
| **Variation graph**         | Reference sequence plus alternative variants    | Represent genetic variation                         |
| **Genome graph**            | Genome sequences and their relationships        | General graph-based genome representation           |
| **Pangenome graph**         | Multiple genomes of a species                   | Represent species/population-wide genetic diversity |



# Question 2: BFS and DFS in Graph-Based Genomic Analysis

## Breadth-First Search (BFS)

**BFS** explores a graph **level by level**, visiting neighboring nodes before moving farther away.

In genomic analysis, BFS can be used to:

* Find the **shortest path** between genomic nodes.
* Explore nearby sequences in **assembly and variation graphs**.
* Identify closely connected genomic regions or variants.

**Example:** BFS can find the shortest path between two nodes representing genomic regions in a genome graph.

## Depth-First Search (DFS)

**DFS** explores one path **as deeply as possible** before backtracking.

In genomic analysis, DFS can be used to:

* Explore possible paths through **assembly and variation graphs**.
* Identify **connected components** in genomic networks.
* Trace alternative genomic sequences or variant paths.

**Example:** DFS can follow one possible sequence path in an assembly graph until it reaches an endpoint, then backtrack to explore another path.

## Key Difference

| BFS                       | DFS                                 |
| ------------------------- | ----------------------------------- |
| Explores level by level   | Explores deeply along one path      |
| Uses a queue              | Uses a stack/recursion              |
| Useful for shortest paths | Useful for exploring possible paths |


# Question 3: Linear and Graph-Based Reference Genomes

## Linear Reference Genome

A **linear reference genome** represents a genome as a single continuous DNA sequence.

* Provides a standard sequence for **read alignment and variant calling**.
* Simple to store, analyze, and interpret.
* Represents only one version of each genomic region.
* May cause **reference bias** because sequences that differ from the reference may be poorly represented.

**Example:** A single reference genome of *Mycobacterium tuberculosis* used to align sequencing reads from different TB isolates.

## Graph-Based Reference Genome

A **graph-based reference genome** represents genomic sequences as **nodes and connections**, allowing multiple alternative sequences to be represented.

* Can represent **genetic variation**, including SNPs, insertions, and deletions.
* Reduces reference bias by allowing reads to align to alternative paths.
* Better represents genetic diversity across multiple individuals or strains.
* More complex to construct and analyze than a linear reference.

**Example:** A pangenome graph containing genomic sequences from multiple *M. tuberculosis* strains.

# Question 4: Construct a De Bruijn Graph

Given the reads:

```text
ATGC
TGCA
GCAT
CATG
```

Using **k = 3**:

## 1. List all 3-mers

Each 4-base read produces two 3-mers:

| Read | 3-mers   |
| ---- | -------- |
| ATGC | ATG, TGC |
| TGCA | TGC, GCA |
| GCAT | GCA, CAT |
| CATG | CAT, ATG |

Therefore, the 3-mers are:

```text
ATG, TGC, TGC, GCA, GCA, CAT, CAT, ATG
```

---

## 2. Identify all 2-mer nodes

For a De Bruijn graph, the nodes are the **(k−1)-mers = 2-mers**.

Unique 2-mers:

```text
AT, TG, GC, CA
```

---

## 3. Draw the Directed De Bruijn Graph

Each 3-mer becomes a directed edge from its first 2 bases to its last 2 bases:

```text
ATG → AT → TG
TGC → TG → GC
GCA → GC → CA
CAT → CA → AT
```

Since each 3-mer occurs twice, each edge occurs twice:

```text
        2
   ┌─────────┐
   ↓         │
  AT ──2──→ TG
   ↑         │
   │         │ 2
   │         ↓
  CA ←─2─── GC
```

Therefore, the graph can be represented as:

```text
AT --2--> TG --2--> GC --2--> CA --2--> AT
```

where **2** means there are two parallel edges.

---

## 4. Calculate In-degree and Out-degree

| Node | In-degree | Out-degree |
| ---- | --------: | ---------: |
| AT   |         2 |          2 |
| TG   |         2 |          2 |
| GC   |         2 |          2 |
| CA   |         2 |          2 |

All nodes have:

```text
In-degree = Out-degree = 2
```

---

## 5. Determine Whether an Eulerian Path or Cycle Exists

An **Eulerian cycle exists** because every node has equal in-degree and out-degree, and the graph is connected.

```text
AT → TG → GC → CA → AT
```

This cycle can be repeated twice to use all 8 edges.

---

## 6. Reconstruct the Sequence

Following the Eulerian cycle and overlapping the nodes:

```text
AT → TG → GC → CA → AT → TG → GC → CA → AT
```

The reconstructed sequence is:

```text
ATGCATGCAT
```

Because the graph forms a repeating cycle, the underlying **circular sequence** can also be represented as:

```text
ATGCATGC
```

### Final Answer

```text
3-mers: ATG, TGC, TGC, GCA, GCA, CAT, CAT, ATG

2-mer nodes: AT, TG, GC, CA

Edges:
AT → TG (2)
TG → GC (2)
GC → CA (2)
CA → AT (2)

All nodes:
In-degree = 2
Out-degree = 2

Eulerian cycle: YES

Reconstructed sequence: ATGCATGCAT
Circular sequence: ATGCATGC
```

# Question 5: Construct a Pangenome Graph

### Given Genomes

```text
Genome 1: ACTGAC
Genome 2: ACTAAC
Genome 3: ACTGGAC
```

## 1. Identify Shared Sequence Regions

The regions **ACT** and **AC** are shared by all three genomes.

```text
Shared: ACT → AC
```

---

## 2. Identify the SNP

At position 4:

```text
Genome 1: ACTGAC
Genome 2: ACTAAC
Genome 3: ACTGGAC
         ↑
```

Genome 2 has **A**, while Genomes 1 and 3 have **G**.

**SNP: G → A**

---

## 3. Identify the Insertion

Genome 3 contains an additional **G**:

```text
Genome 1: ACTGAC
Genome 2: ACTAAC
Genome 3: ACTGGAC
              ↑
```

Therefore, **G is inserted in Genome 3** between the existing G and AC region.

---

## 4. Pangenome Graph

A simplified graph representing all three genomes is:

```text
             ┌── G ── A ──┐
ACT ─────────┤             ├── AC
             │   ┌── G ── G ── A ──┐
             └── A ── A ────────────┘
```

A clearer path representation is:

```text
Genome 1:  ACT → G → A → C
Genome 2:  ACT → A → A → C
Genome 3:  ACT → G → G → A → C
```

---

## 5. Each Genome as a Path

```text
Genome 1: ACT → G → A → C

Genome 2: ACT → A → A → C

Genome 3: ACT → G → G → A → C
```

---

## 6. Branch and Rejoining Nodes

* **Branching node:** `ACT`
* From `ACT`, the genomes follow different paths because of the SNP.
* Genome 3 has an additional branch containing the inserted `G`.
* The paths eventually **rejoin at the shared AC region**.

---

## 7. Classify Graph Segments

| Segment                 | Classification    | Reason                                  |
| ----------------------- | ----------------- | --------------------------------------- |
| **ACT**                 | Core              | Present in all three genomes            |
| **AC**                  | Core              | Present in all three genomes            |
| **G/A variant region**  | Accessory/variant | Different sequences occur among genomes |
| **Extra G in Genome 3** | Unique            | Present only in Genome 3                |

### Final Summary

```text
Core:      ACT, AC
SNP:       G → A in Genome 2
Insertion: Extra G in Genome 3
Branches:  Alternative G/A paths and inserted G path
Unique:    Extra G segment in Genome 3
```

# Question 6: Graph Traversal

### Given Graph

```text
A → B → D
A → C → D
C → E → F
```

## 1. BFS from A

BFS visits nodes **level by level**.

```text
Start: A
Level 1: B, C
Level 2: D, E
Level 3: F
```

**BFS traversal:**

```text
A → B → C → D → E → F
```

---

## 2. DFS from A

Following the graph in the given order:

```text
A → B → D → C → E → F
```

**DFS traversal:**

```text
A → B → D → C → E → F
```

---

## 3. All Paths from A to D

There are **two paths**:

```text
1. A → B → D
2. A → C → D
```

---

## 4. Branching Nodes

A **branching node** has more than one outgoing edge.

* **A** branches to `B` and `C`.
* **C** branches to `D` and `E`.

Therefore:

```text
Branching nodes: A, C
```

---

## 5. Is the Graph Acyclic?

**Yes.** The graph is **acyclic** because there is no path that leads back to a previously visited node.

Therefore, the graph is a **Directed Acyclic Graph (DAG)**.
